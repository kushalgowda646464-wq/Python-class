```python
import pandas as pd
```

```python
#series vs dataframe

#a series is like a single column of data
#it is one dimentional
```

```python
s=pd.Series([10,20,30,40])
print(s)
```

```python
#custom index

s=pd.Series([85,90,88],index=["asha","ravi","kiran"])
print(s)
```

```python
#dataframe
#a dataframe is a table (rows+columns)
#it is 2-dimensional

data={
    "name":["asha","ravi","kiran"],
    "marks":[85,98,67]
}
df=pd.DataFrame(data)
print(df)

print(df.index)
print(df.columns)
print(df.values)
```
