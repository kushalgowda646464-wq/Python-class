#pandas selection

abel-based selection
#.loc

#used when selecting by row lables a#lnd column names

syntax-
df.loc[row_lable,column_lable]



```python
import pandas as pd
df=pd.read_csv("North America Restaurants (1).csv")
df.head()
```

```python
df.loc[0,"name"]
```

```python
df.loc[1]

# to select entire row
```

```python
df.loc[0:2]

#multiple row
#.loc includes end index
```

```python
df.loc[:,["name","city"]]

#to select specific column data
```

#loc
position based selection

used when selecting by position

syntax:
df.iloc[row_index,column_index]

```python
df.iloc[0,1]
#selecting single value
```

```python
df.iloc[0:3,0:2]
#selecting multiple rows and colimns
#iloc excludes end index
```

```python
#boolean masking(filtering)
#filters rows based on condition

df[df["weighted_rating_value"]>2.5]

#singale condition
```

```python
#for multiple conditions

df[(df["weighted_rating_value"]>2.5) & (df["weighted_rating_value"]>3.5)]
```

```python
#or condition 
df[(df["weighted_rating_value"]>4.5) & (df["pickup_enabled"]==False)]
```

```python
#not condition
df[(df['country']=='cA')]
```

```python
#query method 
#similar sql query

df.query('weighted_rating_value>3.6')
```

```python
#multiple condition
df.query('weighted_rating_value>2.6 & country=="CA"')
```

```python
#using variables

limit=4.2
df.query('weighted_rating_value>@limit')

# @ allows python to  use variable inside query
```
