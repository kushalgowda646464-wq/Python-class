```python
#exceptions(error handling)
#try-except

try:
    num=int(input("enter a number:"))
except ValueError:
    print("invalid input")
```

zero division error - divide by zero error
file not found error - file missing
index error- list index wrong
key error-dictionary key missing
attribute error - method dosent exist
import error - module cannot be imported 
exception - catches almost all errors (not recommended for specific error handling)


```python
try:
    print(age)
except NameError:
    print("variable not defined")
```

```python
try:
    print(10/0)
except ZeroDivisionError:
    print("cannot divide by zero")
```

```python
try:
    open("abc.txt","r")
except FileNotFoundError:
    print("file not found")
```

```python
try:
    data={"name":"sam"}
    print(data["age"])
except KeyError:
    print("key not found")
```

```python
#exception is generic, it catches almost all errors
try:
    x=10/0
except Exception:
    print ("somehting wenr qrong")

#not recommended for specific errors
```

```python
#catching multiple errors
try:
    a=int(input())
    b=int(input())
    print(a/b)
except(ValueError,ZeroDivisionError):
    print("invalid input")
```

finally-
finally is a block that always runs whether there is an error or not

basic syntax:
try:
    #rrisky code
except:
    #handle error
finally:
    #alwyas runs

we use finally to:
1.close files
2.close database connections
3.release resources
4.cleanup actions

```python
try:
    print(10/0)
except ZeroDivisionError:
    print("connot divide by zero")
finally:
    print("done")
```

```python
#file handling example

file=None

try:
    file=open("tool.txt","r")
    print(file.read())
except FileNotFoundError:
    print("file not found")
finally:
    if file:
        file.close()
        print("file closed")
```

```python
#with return statement 

def test():
    try:
        return "try block"
    finally:
        print("finally block")

print(test())
```
