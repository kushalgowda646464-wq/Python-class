Loops

loops are used to execute the same block of code multiples time without writing the code repeatdly

there are two types-
for loop
while loop

for loop-
a for loop runs once for every item in a collection

collection=list,string,dictionary,etc

syntax-
for item in iterable:
     action

```python
students=["raj", "sam", "max", "rekha"]

for name in students:
    print(name)
```

```python
genders=["m", "f", "m"]

for g in genders:
    print(g.lower())
```

```python
# with condition
marks=[30,50,90]
 
for m in marks:
    if m>=40:
        print("pass")
    else:
        print("fail")
```

```python
word="data"
for w in word:
    print(w)
    
```

```python
#stop when value found
#stop the loop when 20 is found

nums=[10,11,12,20,30,40]

i=0
while i<len(nums):
    if nums[i]==20:
        print("value found")
        break
    print("number found is:", nums[i])
    i=i+1
```
