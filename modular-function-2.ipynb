Docstring
a string written inside a function to explain what it does


it helps:
other developers
future you
interviewers
documentation tools

it use triple quotes- """  """

```python
def add(a,b):
    """
    adds teo numbers.
    """
    return a+b

add(7,6)
print(add.__doc__)#to acess the docstring written inside
```

```python
#google style docstring
def add(a,b):
    """
    add two numbers.

    args:
    a(int):first numbers
    b(int):second numbers
    
    return:
    int:sum of a and b
    """
    return a+b
```

```python
#numpy style docstring

def add(a,b):
    """
    add two numbers:

    parameters
    ----------
    a:int
      first number.
    b:int
      second number.

    returns
    -------
    int
       sum of a and b
    """
    return  a+b
```

```python
#*args- variable positional arguments
#*args collects extra psitionals arguments into a tuple

#sum all numbers

def total (*numbers):
    total_sum=0
    for n in numbers:
        total_sum+=n
    return total_sum

print(total(10,20,40))
print(total(50,60))
```

```python
def greet(greeting, *names):
    for name in names:
        print(greeting,name)

greet("good morning", "ram","sam","cam")

```

```python
def show_types(*args):
    for value in args:
        print(value,"->",type(value))

show_types(10, "sam", 5.5)
```

```python
#kawargs
#variable keyword arguments

def show_detailes(**kwargs):
    for key,value in kwargs.items():
        print(key,".", value)

show_detailes(name="sam",age=21)
```

```python
def check_mail(**kwargs):
    if "email" in kwargs:
        return "email provided"
    return "email missing"

print(check_mail(name="rahul",email="test@gamil.com"))
```

```python
#update default settings
def setting(**red):
    config={"theme":"light","front":12}
    config.update(red)
    return config

print(setting(theme="dark"))
```

```python
#combined example

def example(a,*args,**kwargs):
    print("a:",a)
    print("args",args)
    print("kwargs",kwargs)

example(1,2,3,4,name="manish",song="retro")
```

lambda(anonymous function)-
lambda=small function written in one line

syntax:
lambda parameters: expression

no return keyword needed.

use lambda for-

short logic
inside map filter
one-time use

```python
#normal function

def square(x):
    return x*x

#lambda function (singal parameter)

square=lambda x:x*x
print(square(5))

#lambda function (multiple parameters)

add=lambda a,b:a+b
print(add(5,3))
```

```python
#check even

is_even=lambda x:x%2==0
print(is_even(6))

# get last character

last_char=lambda word:word[-1]
print(last_char("python"))

#convert to uppercase

upper=lambda word:word.upper()
print(upper("date"))
```

```python
#map()
#applies function to every element 

#normal loop

nums=[1,2,3]
square=[]

for n in nums:
    square.append(n*n)
print(square)


#using map()

nums=[1,2,3]
square=list(map(lambda x:x*x,nums))
print(square)
```

```python
#add 10 to each

nums=[5,10,15]
result=list(map(lambda x:x+10,nums))
print(result)

#uppercase names
names=["amit", "snjana"]
result=list(map(lambda x:x.upper(),names))
print(result)
```

```python
#filter function 
#reuturn only elements that satisfy condition

#normal loop
nums=[1,2,3,4]
evens=[]

for n in nums:
    if n%2==0:
        evens.append(n)
print(evens)

#using filter()

nums=[1,2,3,4]
evens=list(filter(lambda x:x%2==0,nums))
print(evens)
```

```python
#remove empty strings

data=["rahul","","saketh","","soumya"]
result=list(filter(lambda x:x!="",data))
print(result)
```

```python
#reduce 
#used for comulative calculation
#have to import for this
```

```python
#sum of all elements

from functools import reduce
nums=[1,2,3,4]
result=reduce(lambda x,y:x+y,nums)
print(result)
```

```python
#concatenate strings

words=["data","science"]
print(reduce(lambda x,y:x+" "+y,words))
```

```python
#find max

nums=[10,50,20]
print(reduce(lambda x,y:x if x>y else y,nums))
```

#score- legb rule

#scope determines where the variable lives

LEGB stands for:
L-LOCAL
E-ENCLOSING
G-GLOBAL
B-BUILT-IN

```python
#local scope
def add():
    a=5
    b=6
    print(a+b)

add()

#here a,b are local variabales and they exist only inside function
```

```python
#global scope
age=24

def show():
    print(age)

show()
#here age is the global variable
#function can read globale variable
```

```python
#enclosing scope

def outer():
    x=10
    def inner():
        print(x)
    inner()

outer()
#here x is ecclosing scope variable
#inner function uses variable from outer function
#inner function uses variable from outer function
```

```python
#built-in scope

x='names'
print(x)
a=len(x)
print(a)
range(5)

len=5
a=len(x)    #now len function wont work because value of len is overwritten
print(a)
```

```python
#global +local same name
x=100

def show():
    x=50
    print("inside function",x)

show()
print("outside function",x)
#local variable does not change or effect global variable
```

```python
#using global variable inside function

total=0

def add_amount(amount):
    global total
    total+=amount

add_amount(3000)
add_amount(500)
print(total)
```
