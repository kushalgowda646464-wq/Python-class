```python
import pandas as pd
import numpy as np

data={
    'customer_name':['asha','bala','chitra','deepa','esha','bala'],
    'city':['bangalore','delhi','mumbai','delhi','bangalore','delhi'],
    'order_value':[5000,7000,np.nan,6500,7200,7000],
    'email':['asha123@gmail.com','bala@gmail.com','chitra@company.com',
             'deepa_98@yahoo.com','esha@gmail.com','bala@gmail.com']
}

df=pd.DataFrame(data)
df
```

Handling duplicate values

Duplicates occur when the same record appears multiple times in a dataset

common reason:
data entry errors
merging datasets
system logging errors

Duplicate can dustrot the analysis, for example:
revenue may appear higher than reality

```python
df.duplicated()
```

```python
#viewing duplicates rows
df[df.duplicated()]
```

```python
#removes comletley identical rows
df.drop_duplicates(inplace=True)
df
```

```python
#using subset

df1=df.drop_duplicates(subset='email')
df1

# this removes duplicates only onthe email column
```

```python
#keep decides which duplicate record remains

df1.drop_duplicates(keep=False)
df1

# keep parameter accepts only three values-first,last,false
```

common regex patterns

\d          digit
\w          latter or number
^           start of a string
$           end of string



commonly used with-
.str.contains()
.str.extract()
.str.replace()
.str.match()

```python
import pandas as pd

data={"email":["asha123@gmail.com",
                  "bala@gamil.com",
                  "chithra@gmail.com",
                  "deepa_98@gamil.com"]}
df=pd.DataFrame(data)
```

```python
#extract numbers from text

df["email"].str.extract(r"(\d+)")
```

```python
#find rows containing numbers

df["email"].str.contains(r"\d")
```

```python
#extract email domain
df["email"].str.extract(r"@(\w+)")
```

when we write @(\w+)

() capture group
\w letter or number
+ one or more

here match strats at the symbols @ but since @ is outside of the capture group it is not extract
so only what is inside the capture group () is extracted

```python
#find emails ending with gmail.com

df["email"].str.contains(r"@gmail\.com$")
```

```python
data={"Name":["asha","bala","chitra","deepa"]}
df=pd.DataFrame(data)
```

```python
#find names sarting with A
df["Name"].str.contains(r"^A")
```

```python
#remove numbers from emails

df["email"].str.replace(r"\d","",regex=True)
```

```python
#regex groups
#deepa_98@gmail.com
#we want to extract both and number in the email separetly

df["email"].str.extract(r"(\w+)_(\d+)")
```
