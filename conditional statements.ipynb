match-case

match-value:
   case option1:
   action
   case option2:
    case-:
        defult

```python
role="manager"

match role:
    case "admin":
        print("full access")
    case "manger":
        print("limited access")
    case _:
        print("access to only view")
    
```

```python
#ticket status
status="p"

match status:
    case "p":
        print("pending")
    case "c":
        print("waiting")
    case "r":
        print("pending")
    case "w":
        print("pending")
    case _:
        print("unknown")
```

```python
city="delhi"

match city:
    case "delhi"|"bangalore":
        print("metro city")
    case _:
        print("tier 2 city")
    
```
