#data transformation

.map() replaces values using a dictionary mapping

map in pandas is used to transform or substiture values ina singale column(series)
using a mapping rule.
commanly used in data cleaning or feature engineering

syntax-series.map(mapping)

here mapping can be another:

dictionary
a function
another series

```python
import pandas as pd
data={"city":
    ["bangalore","delhi","mumbai"]
}
df=pd.DataFrame(data)
df
```

```python
#convert city names to regions
region_map={"bangalore":"south",
             "delhi":"north",
             "mumbai":"west"
              } 

df["region"]=df["city"].map(region_map)
df

```

```python
#mapping using a function

#.map() can apply a function to each value

df["city_length"]=df["city"].map(len)
df
```

```python
#mapping through a series
city_series=pd.Series({"bangalore":"south",
                       "delhi":"north",
                       "mumbai":"west"})

df["region"]=df["city"].map(city_series)
df
```

```python
#mapping with missing values
#when there are missing values its filled 
region_map={"bangalore":"south",
             "delhi":"north",
            }

df["region"]=df["city"].map(region_map)
df
```

```python
#filling missing values using map

df["region"]=df["city"].map(region_map).fillna("other")
df
```

```python
import pandas as pd
import numpy as np 

data={"customer":["asha","bala","chitra"],
      "sales":[50000,30000,15000]}

df=pd.DataFrame(data)
```

```python
#Applying custom functions
#.apply() lets us run custom logic onrows or columns

#example -classify slaes performance

def sales_category(x):
    if x>40000:
        return "high"
    elif x>25000:
        return "medium"
    else:
        return "low"
    
# apply function

df["sales_category"]=df["sales"].apply(sales_category)
df
```

```python
#apply with axis 

#axis=0            oparate on column
#axis= 1           operate on rows

df.apply(np.sum,axis=0)
# calculates sum colimn -wise
```

```python
data={
    "a":[10,20,30,40],
    "b":[1,2,3,4]
}

df=pd.DataFrame(data)
```

```python
df.apply(lambda row:row["a"]+row["b"],axis=1)

#row wise operation
```

```python
#create a column called double_sales using lamba with apply function

df["Double_sales"]=df["sales"].apply(lambda x:x*2)
df
```

```python
#apply on entire dataframe

df.apply(lambda x:x*2)
```

```python
data={"revenue":[50000,30000],
       "cost":[20000,15000]
       }
df=pd.DataFrame(data)
df
```

```python
df["profit"]=df.apply(lambda row:row["revenue"]-row["cost"],axis=1)
df
```
