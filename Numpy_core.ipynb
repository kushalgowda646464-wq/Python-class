#brodcasting

#broadcasting allows operations between arrays of different shapes

#broad casting eules-
two arrays are compatible if:

1. they have same shape
   or

2. one of them is 1
   or

3, smaller array can stretch to match larger

```python
arr=np.array([1,2,3])

print(arr+5)
```

```python
#broadcasting 1D with 2D

arr=np.array([[1,2,3],
             [4,5,6]])

b=np.array([10,20,30])

print(arr+b)
```

```python
a=np.array([[1,2,3],
           [4,5,6]])
b=np.array([1,2])
print(a+b)

#broadcasting fails
```

```python
#column braodcasting

arr=np.array([[1,2,3],
              [4,5,6]])

col=np.array([[10],
            [20]])
print(arr+col)
```

#vectorization
doing operations on entire array at once instead of using loops

vectiorization makes it:
-faster
-cleaner code
-less memory usage


```python
#square all numbers
arr=np.array([1,2,3,4])
print(arr**2)
```

```python
a=np.array([1,2,3])
b=np.array([4,5,6])

print(a+b)
```

```python
#conditional vectorization

arr=np.array([10,20,30,40])
print(arr>25)
```

```python
#using condition to modify values

arr[arr>25]=100
print(arr)
```

```python
#using where

np.where(arr>25,1,0)
```

```python
#mathematical operations

arr=np.array([1,2,3,4])

print(np.sort(arr))
print(np.sqrt(arr))
print(np.log(arr))
print(np.sin(arr))
```

```python
#braodcasting +vectorization together

arr=np.array([[1,2,3],
              [4,5,6]])

mean=np.array([2,3,4])

print(arr-mean)

#broadcasting handles shapes, vectorization handles operation
```
